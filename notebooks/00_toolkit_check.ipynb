### Raster: make a tiny GeoTIFF, summarize, and plot

```python
import numpy as np, rasterio
from rasterio.plot import show
from rasterio.transform import from_origin

demo = 'demo.tif'
with rasterio.open(demo, 'w', driver='GTiff', height=32, width=32, count=3,
                   dtype='uint8', crs='EPSG:4326', transform=from_origin(0, 32, 1, 1)) as dst:
    dst.write(np.random.randint(0, 255, (3, 32, 32), dtype='uint8'))

with rasterio.open(demo) as src:
    print('bands', src.count, '| size', src.width, 'x', src.height, '| crs', src.crs)
    show(src)
```

### Vector: load a real dataset via geodatasets (replaces the removed geopandas.datasets)

```python
import geopandas as gpd
from geodatasets import get_path

nybb = gpd.read_file(get_path('nybb'))
print(nybb.crs, nybb.shape)
nybb.plot(column='BoroName', legend=True, figsize=(7, 6))
```
