### Tested helpers (the numeric core you build on)


```python
import numpy as np
def velocity_fit(times_years, disp):
    t = np.asarray(times_years, float); d = np.asarray(disp, float)
    m = ~(np.isnan(t) | np.isnan(d))
    if m.sum() < 2: return float("nan"), float("nan")
    a = np.vstack([t[m], np.ones(m.sum())]).T
    slope, intercept = np.linalg.lstsq(a, d[m], rcond=None)[0]
    return float(slope), float(intercept)
def los_to_vertical(los, incidence_deg):
    return np.asarray(los, float) / np.cos(np.deg2rad(np.asarray(incidence_deg, float)))
```

### Self-check (these mirror the module's tests)


```python
rng = np.random.default_rng(0); t = np.linspace(2019, 2025, 60)
d = -18.0*(t-t[0]) + rng.normal(0, 2, t.size)
assert abs(velocity_fit(t, d)[0] + 18.0) < 1.0
assert abs(los_to_vertical(1.0, 60.0) - 2.0) < 1e-9
print("ok")
```
