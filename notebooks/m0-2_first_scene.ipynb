### Tested helper


```python
def cloud_query(max_cloud):
    "STAC property filter for maximum cloud cover (pure, testable)."
    return {"eo:cloud_cover": {"lt": max_cloud}}
```

### Self-check


```python
assert cloud_query(20) == {"eo:cloud_cover": {"lt": 20}}
print("ok")
```

### Find a real scene (edit the bbox and dates for your AOI)


```python
from pystac_client import Client
STAC = "https://stac.dataspace.copernicus.eu/v1"
bbox = [3.3, 6.4, 3.5, 6.6]   # Lagos area; change to your AOI
client = Client.open(STAC)
search = client.search(collections=["sentinel-2-l2a"], bbox=bbox,
                       datetime="2025-01-01/2025-03-31", query=cloud_query(20))
items = list(search.items())
print(len(items), "scenes found")
if items:
    it = items[0]
    print("date:", it.datetime, "| cloud:", it.properties.get("eo:cloud_cover"))
    print("bands:", sorted(it.assets)[:12])
```
