### Tested helpers (the numeric core you build on)


```python
import numpy as np
def log_ratio(pre, post, eps=1e-10):
    "Log-ratio change on linear-power SAR; large magnitude = change."
    pre = np.maximum(np.asarray(pre, float), eps); post = np.maximum(np.asarray(post, float), eps)
    return np.log(post / pre)
```

### Self-check (these mirror the module's tests)


```python
lr = log_ratio(np.array([1.0, 1.0]), np.array([1.0, 4.0]))
assert abs(lr[0]) < 1e-9 and lr[1] > 0
print("ok")
```
