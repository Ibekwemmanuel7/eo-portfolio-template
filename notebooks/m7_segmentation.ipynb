### Tested helpers (the numeric core you build on)


```python
import numpy as np
def iou_score(pred, target, eps=1e-6):
    pred = np.asarray(pred, bool); target = np.asarray(target, bool)
    inter = (pred & target).sum(); union = (pred | target).sum()
    return float((inter + eps) / (union + eps))
def tile_indices(H, W, size, stride):
    def axis(n):
        if n <= size: return [0]
        idx = list(range(0, n - size + 1, stride))
        if idx[-1] != n - size: idx.append(n - size)
        return idx
    return [(y, x) for y in axis(H) for x in axis(W)]
```

### Self-check (these mirror the module's tests)


```python
a = np.array([[1,1],[0,0]]); b = np.array([[1,0],[0,0]])
assert abs(iou_score(a, b) - 0.5) < 1e-3 and iou_score(a, a) > 0.99
assert (0,0) in tile_indices(10,10,5,5) and len(tile_indices(10,10,5,5)) == 4
print("ok")
```
