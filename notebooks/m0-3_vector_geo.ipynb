### Tested helper: pick the right UTM zone for area and distance


```python
def utm_epsg(lon, lat):
    "EPSG code of the UTM zone for a lon/lat (area and length must be metric)."
    zone = int((lon + 180) // 6) + 1
    return (32600 if lat >= 0 else 32700) + zone
```

### Self-check


```python
assert utm_epsg(3.4, 6.5) == 32631   # Lagos: UTM 31N
assert utm_epsg(151.2, -33.9) == 32756  # Sydney: UTM 56S
print("ok")
```

### Load a vector, inspect its CRS, and compute area correctly


```python
import geopandas as gpd
from geodatasets import get_path
nybb = gpd.read_file(get_path('nybb'))
print('CRS:', nybb.crs, '| rows:', len(nybb))
# already projected (EPSG:2263, feet); area in a metric CRS:
area_km2 = nybb.to_crs(32618).area / 1e6
print(area_km2.round(1).tolist())
nybb.plot(column='BoroName', legend=True, figsize=(7, 6))
```
