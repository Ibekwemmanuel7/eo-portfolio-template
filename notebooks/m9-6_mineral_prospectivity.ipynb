### Tested helpers (the numeric core you build on)


```python
import numpy as np
def band_ratio(numer, denom, eps=1e-6): return np.asarray(numer, float) / (np.asarray(denom, float) + eps)
def spectral_angle(a, b, eps=1e-12):
    a = np.asarray(a, float); b = np.asarray(b, float)
    num = float(np.dot(a, b)); den = float(np.linalg.norm(a)*np.linalg.norm(b)) + eps
    return float(np.arccos(np.clip(num/den, -1, 1)))
def woe_contrast(n_total, n_deposit, n_evidence, n_both):
    n_total,n_deposit,n_evidence,n_both = map(float,(n_total,n_deposit,n_evidence,n_both))
    wp = np.log((n_both/n_deposit)/((n_evidence-n_both)/(n_total-n_deposit)))
    wm = np.log(((n_deposit-n_both)/n_deposit)/(((n_total-n_evidence)-(n_deposit-n_both))/(n_total-n_deposit)))
    return float(wp), float(wm), float(wp-wm)
```

### Self-check (these mirror the module's tests)


```python
assert abs(band_ratio(4.0, 2.0) - 2.0) < 1e-3
a = np.array([0.1,0.2,0.3,0.4]); assert spectral_angle(a, 5*a) < 1e-4
assert abs(woe_contrast(1000,100,200,80)[2] - 3.2580966) < 1e-4
print("ok")
```
