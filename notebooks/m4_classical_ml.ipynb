### Tested helpers (the numeric core you build on)


```python
import numpy as np
def assign_blocks(lons, lats, block_deg=0.1):
    bx = np.floor(np.asarray(lons, float) / block_deg).astype(int)
    by = np.floor(np.asarray(lats, float) / block_deg).astype(int)
    return np.array([f"{a}_{b}" for a, b in zip(bx, by)])
```

### Self-check (these mirror the module's tests)


```python
b = assign_blocks([7.501, 7.509, 7.62], [5.001, 5.004, 5.10], 0.1)
assert b[0] == b[1] and b[0] != b[2]
print("ok")
```
