### Tested helpers (the numeric core you build on)


```python
import numpy as np
def to_db(x, eps=1e-10): return 10.0 * np.log10(np.maximum(x, eps))
def from_db(d): return 10.0 ** (d / 10.0)
def enl(patch):
    p = np.asarray(patch, float); s = p.std()
    return float("inf") if s == 0 else float((p.mean() / s) ** 2)
def vh_vv_ratio(vh, vv, eps=1e-10): return vh / np.maximum(vv, eps)
```

### Self-check (these mirror the module's tests)


```python
x = np.array([0.5, 1.0, 2.0])
assert np.allclose(from_db(to_db(x)), x, rtol=1e-6)
assert abs(float(to_db(np.array(10.0))) - 10.0) < 1e-9
print("ok")
```
