### Tested helpers (phenology from a season)


```python
import numpy as np
def ndvi(nir, red, eps=1e-9):
    nir=np.asarray(nir,float); red=np.asarray(red,float); return (nir-red)/(nir+red+eps)
def gap_fill(y):
    y=np.asarray(y,float).copy(); idx=np.arange(y.size); g=~np.isnan(y)
    if g.sum()==0: return y
    y[~g]=np.interp(idx[~g], idx[g], y[g]); return y
def harmonic_fit(t, y):
    t=np.asarray(t,float); y=np.asarray(y,float); m=~(np.isnan(t)|np.isnan(y))
    if m.sum()<3: return float('nan'),float('nan'),float('nan')
    w=2*np.pi; A=np.vstack([np.ones(m.sum()), np.sin(w*t[m]), np.cos(w*t[m])]).T
    a,b,c=np.linalg.lstsq(A,y[m],rcond=None)[0]; return float(a),float(np.hypot(b,c)),float(np.arctan2(c,b))
def season_amplitude(y):
    y=np.asarray(y,float); return float(np.nanmax(y)-np.nanmin(y)) if not np.all(np.isnan(y)) else float('nan')
```

### Self-check


```python
assert abs(float(ndvi(np.array(0.4),np.array(0.1)))-0.6)<1e-6
assert np.allclose(gap_fill(np.array([0.0,np.nan,2.0,np.nan,4.0])),[0,1,2,3,4])
t=np.linspace(0,1,48,endpoint=False); y=0.4+0.5*np.sin(2*np.pi*t)
assert abs(harmonic_fit(t,y)[1]-0.5)<0.02
print('ok')
```
