### Tested helpers (the numeric core you build on)


```python
import numpy as np
def cloud_query(max_cloud):
    "STAC property filter for max cloud cover (pure, testable)."
    return {"eo:cloud_cover": {"lt": max_cloud}}

_BAD_SCL = (3, 8, 9, 10, 11)  # shadow, cloud med/high, cirrus, snow
def scl_mask(scl, bad=_BAD_SCL):
    "Boolean keep-mask from a Sentinel-2 SCL array."
    scl = np.asarray(scl)
    return ~np.isin(scl, bad)
```

### Self-check (these mirror the module's tests)


```python
assert cloud_query(20) == {"eo:cloud_cover": {"lt": 20}}
m = scl_mask(np.array([[4, 8], [3, 5]]))
assert m.tolist() == [[True, False], [False, True]]
print("ok")
```
